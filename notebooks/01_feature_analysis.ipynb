```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# load cleaned dataset

df = pd.read_csv("../data/creditcard_cleaned.csv")

# display first 5 rows

df.head()
```

```python
print("Dataset shape: ", df.shape)
print("Dataset columns: ", df.columns.tolist())
```

```python
# select numerical feature column
numerical_features = df.drop(columns=["Class"])

print("Number of numerical features: ", len(numerical_features.columns))
print("\nNumerical features: ", numerical_features.columns.tolist())
```

```python
# calculate correlation between numerical features
correlation_matrix = numerical_features.corr()
# display correlation matrix
correlation_matrix
```

```python
# set figure size
plt.figure(figsize=(16, 12))

# create correlation heatmap
sns.heatmap(correlation_matrix, cmap="coolwarm", center=0,linewidths=0.5)

# add title
plt.title("Correlation heatmap of transaction features")

# display plot
plt.show()
```

```python
# calculate correlation of all features with class
class_correlation = df.corr()["Class"].drop("Class")

# sort correlations from highest to lowest
class_correlation = class_correlation.sort_values(ascending=False)

# display results
print("Feature correlation with fraud: ", display(class_correlation))
```

```python
plt.figure(figsize=(12, 8))

class_correlation.plot(kind="bar")

plt.title("Feature correlation with fraud")
plt.xlabel("Features")
plt.ylabel("Correlation with Class")
plt.axhline(y=0, color="black", linewidth=0.8)

plt.tight_layout()
plt.show()
```

```python
# compare feature averages for legitimate and fraudulent transactions
class_means = df.groupby("Class").mean()

print(class_means)
# display selected features
selected_features = ["V11", "V4", "V2", "V17", "V14", "V12", "Amount"]

display(class_means[selected_features].T)
```

```python
# select important features
selected_features =  ["V11", "V4", "V2", "V17", "V14", "V12", "Amount"]

# create subplots
fig, axes = plt.subplots(2, 4, figsize=(18, 10))

axes = axes.flatten()

# create boxplots

for i, features in enumerate(selected_features):
    sns.boxplot(data=df, x = "Class", y=features, ax=axes[i])

    axes[i].set_title(f"{features} Distribution")
    axes[i].set_xlabel("Class (0 = Legitimate, 1 = Fraud)")

# remove the unused subplot
fig.delaxes(axes[7])

plt.tight_layout()
plt.show()
```

```python
# selected features for comparison
selected_features = [    "V11", "V4", "V2",
    "V17", "V14", "V12", "Amount"]

# calculate median and quartiles for each class
for feature in selected_features:
    print(f"\nFeature: {feature}")
    display(df.groupby("Class")[feature]
            .describe()[["25%", "50%", "75%"]])
```
